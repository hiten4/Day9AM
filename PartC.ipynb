Q1. Shallow Copy vs Deep Copy

Shallow Copy

A shallow copy creates a new outer object but does not copy the nested objects inside it.
Instead, the inner objects are still referenced from the original list.

Example:

import copy

original = [[1, 2], [3, 4]]
shallow = copy.copy(original)

shallow[0][0] = 100

print(original)
print(shallow)

Output:

[[100, 2], [3, 4]]
[[100, 2], [3, 4]]

Both lists changed because the inner lists are shared.

Why shallow copy fails

Shallow copy only copies the first level of the object.
Nested mutable objects like lists still point to the same memory location.

Deep Copy

A deep copy creates a completely independent copy of the object, including all nested objects.

Example:

import copy

original = [[1, 2], [3, 4]]
deep = copy.deepcopy(original)

deep[0][0] = 100

print(original)
print(deep)

Output:

[[1, 2], [3, 4]]
[[100, 2], [3, 4]]

The original list remains unchanged.

When deep copy is required

Deep copy is required when working with nested data structures and you want to modify the copied structure without affecting the original data.

Q2. Coding — List Rotation

```python
def rotate_list(lst, k):

    n = len(lst)

    if n == 0:
        return lst

    k = k % n

    return lst[-k:] + lst[:-k]


print(rotate_list([1,2,3,4,5], 2))
```

Q3. Debug Problem

Buggy code:

nums = [1, 2, 3, 4, 5, 6, 7, 8]

for num in nums:
    if num % 2 == 0:
        nums.remove(num)

print(nums)

Why the bug happens

The list is being modified while it is being iterated.
When an element is removed, the list shifts left and the loop skips the next element.

Example:

[2,4,6,8]

After removing 2, the list becomes [4,6,8], but the loop moves to the next index and skips 4.

```python
#Correct Solution

nums = [1,2,3,4,5,6,7,8]

nums = [n for n in nums if n % 2 != 0]

print(nums)
```
